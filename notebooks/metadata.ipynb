```python
foldername = "../data/testing1/"

filename = os.listdir(foldername)
```

```python
modulo_info_files = []
for file in filename:
    if "modulo" in file:
        modulo_info_files.append(file)
```

```python
modulo_info = pd.read_csv(os.path.join(foldername,modulo_info_files[0]), header = None)
modulo_info
```

```python
for modulo_info_filename in modulo_info_files:
        modulo_info = pd.read_csv(os.path.join(foldername,modulo_info_filename), header = None)
        if len(modulo_info.columns) > 0:
            modulo_info.drop(columns=modulo_info.columns[5:39], inplace=True)
            modulo_info.columns = ['model_confidence', 1, 2, 3, 4, 'zaber_radius', 'loudness_value', 41, 42, 'animal_name', 'experiment_type', 'timestamp']
        else:
            print("number of columns in the modulo_info file is{}".format(len(modulo_info.columns)))
        #assign the animal name to a variable
        print(modulo_info.animal_name[0])
        animal_name = list(modulo_info.animal_name)[0]
        print(animal_name)
        animal_name = animal_name.lower()
        print(animal_name)
        experiment_type = list(modulo_info.experiment_type)[0]
        print(experiment_type)
        zaber_radius = list(modulo_info.zaber_radius)[0]
        print(zaber_radius)
        
        file_base = os.path.basename(modulo_info_filename)
        print(file_base)

        #cut out the rest of the details and keep only datetime
        file_datetime = file_base[12:-4]
        print(file_datetime)
       
        if experiment_type == 'crickets':
            experiment_type = '1.6.8.0.444444444444444.0000000000000000.0000000000000000'
        else:
            print("experiment_type_unknown")
```

```python
from datetime import datetime

# Define the variables
var1 = animal_name
var2 = experiment_type
var3 = zaber_radius


# Join the variables with "_"
joined_string = f"{var1}_{var2}_{var3}"

# Get the current time as a string
time_created = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

# Prepare the content for the file
file_content = f"{joined_string}\nTime Created: {time_created}"

# Define the folder and file name
folder_path = foldername
file_name = file_datetime + '_' + animal_name +'.txt'
file_path = os.path.join(folder_path, file_name)

# Ensure the folder exists
os.makedirs(foldername, exist_ok=True)

if os.path.isfile(os.path.join(foldername, file_name)):
    # Open the file in append mode and add the line "xyz"
    # Open the file in append mode
    with open(file_path, "a") as file:
        # Write new lines to the file
        file.write("\nNew line added to the file.\n")
        # Add the date and time of the update
        file.write(f"Updated on: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
        file.write(f"{joined_string}\n")
else:
    # Save the content to the file
    file_path = os.path.join(folder_path, file_name)
    with open(file_path, "w") as file:
        file.write(file_content)



print(f"File saved at: {file_path}")
```
